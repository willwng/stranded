```python
will_strands = np.load('me_2.npy')
will_strands = will_strands[::10]

sim=CLSimulator(strands=will_strands, timestep=0.08, mass=1.0)
sim.create_field()
sim.run(n_steps=10)
updated_strands = sim.get_strands()
```

```python
def plot_cls(poses):
    """
    Creates an interactive visualization of multiple centerlines.

    Parameters:
    - poses: [(n_sites, 3)] Centerline positions
    """
    # Create figure
    fig = go.Figure()

    # Plot each helix
    for idx, pos in enumerate(poses):
        # Plot centerline
        fig.add_trace(go.Scatter3d(
            x=pos[:, 0], y=pos[:, 1], z=pos[:, 2],
            mode='lines',
            line=dict(width=3),
            name=f'Centerline {idx + 1}'
        ))


    # Update layout
    fig.update_layout(
        scene=dict(
            # xaxis=dict(title='X', range=[-3, 9]),
            # yaxis=dict(title='Y', range=[-3, 9]),
            # zaxis=dict(title='Z', range=[-1, 21]),
            # aspectmode='manual',
            # aspectratio=dict(x=1, y=1, z=1)
            xaxis=dict(title='X'),
            yaxis=dict(title="Y"),
            zaxis=dict(title='Z'),
            aspectmode='cube'
        ),
        title=dict(text='CL Visualization', y=0.95, x=0.5, xanchor='center', yanchor='top'),
    )

    fig.show()
```

```python
plot_cls(updated_strands)
```
